```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class LinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
    def addLast(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
        else:
            self.tail.next=NN
            self.tail=NN
    def addFirst(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
        else:
            NN.next=self.head
            self.head=NN
    def printData(self):
        temp=self.head
        while temp!=None:
            print(temp.data,end=" ")
            temp=temp.next

obj=LinkedList()
obj.addLast(1)
obj.addLast(2)
obj.addLast(3)
obj.addLast(4)
obj.printData()
```
