```python
class Solution:
    def transpose(self, matrix):
        n=len(matrix)
        m=len(matrix[0])
        tMat=[]
        for i in range(m):
            l=[0]*n
            tMat.append(l)
        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                tMat[j][i]=matrix[i][j]
        return tMat
sol = Solution()
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

transposed_matrix = sol.transpose(matrix)

print(transposed_matrix)
```

```python
class Solution:
    def searchMatrix(self, matrix, target):
      n = len(matrix)
      m = len(matrix[0])
      i = 0
      j = m - 1
      while i < n and j >= 0:
          if matrix[i][j] == target:
              return True
          elif target > matrix[i][j]:
              i += 1
          else:
              j -= 1
      return False
```

```python
class Solution:
    def shiftGrid(self, grid, k):
        l=[]
        n=len(grid)
        m=len(grid[0])
        for i in range(n):
            for j in range(m):
                l.append(grid[i][j])
        for i in range(k):
            data=l.pop()
            l.insert(0,data)
        idx=0
        for i in range(n):
            for j in range(m):
                grid[i][j]=l[idx]
                idx+=1
        return grid
sol = Solution()
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
k = 1
result = sol.shiftGrid(grid, k)
print(result)
```

```python
class Solution:
    def construct2DArray(self, original, m, n):
        if m*n!=len(original):
            return []
        grid=[]
        for i in range(m):
            l=[0]*n
            grid.append(l)
        idx=0
        for i in range(m):
            for j in range(n):
                grid[i][j]=original[idx]
                idx+=1
        return grid
```
