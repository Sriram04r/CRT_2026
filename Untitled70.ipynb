**Circular Linked List** - The last node is point to the first like Traffic signals.

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class CircularLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
    def addLast(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
            self.tail.next=self.head
        else:
            self.tail.next=NN
            self.tail=NN
            self.tail.next=self.head
    def addfirst(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
            self.tail.next=self.head
        else:
            NN.next=self.head
            self.head=NN
            self.tail.next=self.head
    def printData(self):
        temp=self.head
        while temp!=self.tail:
            print(temp.data,end=" ")
            temp=temp.next
        print(temp.data)
    def getSize(self):
        c=0
        temp=self.head
        while temp!=self.tail:
            c+=1
            temp=temp.next
        return c+1
    def getIthNode(self,idx):
        i=0
        temp=self.head
        while i!=idx:
            i+=1
            temp=temp.next
        return temp
    def addAtMiddle(self,idx,elem):
        NN=Node(elem)
        t1=self.getIthNode(idx-1)
        t2=t1.next
        t1.next=NN
        NN.next=t2
    def deleteAtMiddle(self,idx):
        temp=self.getIthNode(idx-1)

        temp.next=temp.next.next
    def delFirst(self):
        if self.head==None:
            print("Linked list is empty cant perform delete operation")

        elif self.head.next==None:
            self.head=None
            self.tail=None
        else:

            self.head=self.head.next
            self.tail.next=self.head
    def delLast(self):
        if self.head==None:
            print("can't perform del operation")
        elif self.head==None:
            self.head=None
            self.tail=None
        else:
            temp=self.head
            while temp.next!=self.tail:
                temp=temp.next
            temp.next=self.head
            self.tail=temp
obj=CircularLinkedList()
obj.addLast(5)
obj.addLast(2)
obj.addLast(8)
obj.addLast(3)
obj.addLast(0)
obj.delLast()
obj.delFirst()
obj.addAtMiddle(1,5)
obj.deleteAtMiddle(2)
obj.printData()
```

**Double Linked List** - It can follows front and backward connections.

```python

class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
class DbLinkedList:
    def __init__(self):
        self.head=None
        self.tail=None
    def addLast(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
        else:
            self.tail.next=NN
            NN.prev=self.tail
            self.tail=NN
    def addFirst(self,data):
        NN=Node(data)
        if self.head==None and self.tail==None:
            self.head=NN
            self.tail=NN
        else:
            NN.next=self.head
            self.head.prev=NN
            self.head=NN
    def printData(self):
        temp=self.head
        while temp!=None:
            print(temp.data,end=" ")
            temp=temp.next

    def printRevData(self):
        temp=self.tail
        while temp!=None:
            print(temp.data,end=" ")
            temp=temp.prev
    def getSize(self):
        c=0
        temp=self.head
        while temp!=None:
            c+=1
            temp=temp.next
        return c
    def getIthNode(self,idx):
        i=0
        temp=self.head
        while i!=idx:
            i+=1
            temp=temp.next
        return temp
    def addAtMiddle(self,idx,elem):
        NN=Node(elem)
        t1=self.getIthNode(idx-1)
        t2=t1.next
        t1.next=NN
        NN.prev=t1
        NN.next=t2
        t2.prev=NN
    def deleteAtMiddle(self,idx):
        t1=self.getIthNode(idx-1)
        t2=t1.next.next
        t1.next=t2
        t2.prev=t1
    def delFirst(self):
        if self.head==None:
            print("Linked list is empty cant perform delete operation")

        elif self.head.next==None:
            self.head=None
            self.tail=None
        else:

            self.head=self.head.next
            self.head.prev=None
    def delLast(self):
        if self.head==None:
            print("can't perform del operation")
        elif self.head==None:
            self.head=None
            self.tail=None
        else:
            self.tail=self.tail.prev
            self.tail.next=None
obj=DbLinkedList()
obj.addLast(5)
obj.addLast(2)
obj.addLast(8)
obj.addLast(3)
obj.addFirst(0)
obj.addAtMiddle(3,10)
obj.deleteAtMiddle(1)
obj.printData()
print()
obj.printRevData()
```
