1. Left View of Binary Tree

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class Solution:
    def leftView(self, root):
        if root==None:
            return []

        q=[]
        q.append(root)
        l=[]
        while len(q)!=0:
            size=len(q)
            for i in range(size):
                curr=q.pop(0)
                if i==0:
                    l.append(curr.data)
                if curr.left!=None:
                    q.append(curr.left)
                if curr.right!=None:
                    q.append(curr.right)
        return l
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.right = Node(4)

sol = Solution()
print("Left View:", sol.leftView(root))
```

2. Right View of Binary Tree

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
class Solution:
    def rightView(self, root):
        if root==None:
            return []

        q=[]
        q.append(root)
        l=[]
        while len(q)!=0:
            size=len(q)
            for i in range(size):
                curr=q.pop(0)
                if i==size-1:
                    l.append(curr.data)
                if curr.left!=None:
                    q.append(curr.left)
                if curr.right!=None:
                    q.append(curr.right)
        return l
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.right = Node(4)

sol = Solution()
print("Right View:", sol.rightView(root))
```

3. Leaf Similar Trees

```python
from typing import Optional

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def DFS(self, root, list_leaves):
        if root is None:
            return
        if root.left is None and root.right is None:
            list_leaves.append(root.val)
            return
        else:
            self.DFS(root.left, list_leaves)
            self.DFS(root.right, list_leaves)

    def leafSimilar(self, root1: Optional[TreeNode], root2: Optional[TreeNode]) -> bool:
        list1 = []
        list2 = []
        self.DFS(root1, list1)
        self.DFS(root2, list2)
        return list1 == list2

root1 = TreeNode(3)
root1.left = TreeNode(5)
root1.right = TreeNode(1)
root1.left.left = TreeNode(6)
root1.left.right = TreeNode(2)
root1.left.right.left = TreeNode(7)
root1.left.right.right = TreeNode(4)
root1.right.left = TreeNode(9)
root1.right.right = TreeNode(8)


root2 = TreeNode(3)
root2.left = TreeNode(5)
root2.right = TreeNode(1)
root2.left.left = TreeNode(6)
root2.left.right = TreeNode(7)
root2.right.left = TreeNode(4)
root2.right.right = TreeNode(2)
root2.right.right.left = TreeNode(9)
root2.right.right.right = TreeNode(8)

sol = Solution()
print("Are leaf sequences similar? (root1, root2)", sol.leafSimilar(root1, root2))


root3 = TreeNode(1)
root3.left = TreeNode(2)
root3.right = TreeNode(3)

root4 = TreeNode(1)
root4.left = TreeNode(3)
root4.right = TreeNode(2)

print("Are leaf sequences similar? (root3, root4)", sol.leafSimilar(root3, root4))
```

4. Level of Traversal using Trres and Dictionary

```python
class Solution:
    def topView(self, root):
        q=[]
        d={}
        l=[]
        q.append((root,0))

        while len(q)!=0:
            size=len(q)
            for i in range(size):
                t=q.pop(0)
                idx=t[1]
                data=t[0].data
                if idx not in d:
                    d[idx]=data

                if t[0].left!=None:
                    q.append((t[0].left,idx-1))
                if t[0].right!=None:
                    q.append((t[0].right,idx+1))
        maxx=max(d)
        minn=min(d)
        for i in range(minn,maxx+1):
            l.append(d[i])
        return l
```
