1. Pre - Order problem in Leet code

```python
from typing import Optional, List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def POTWL(self, root: Optional[TreeNode], list: List[int]):
        if root is None:
            return
        list.append(root.val)
        self.POTWL(root.left, list)
        self.POTWL(root.right, list)

    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        result = []
        self.POTWL(root, result)
        return result

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)

sol = Solution()
result = sol.preorderTraversal(root)

print(result)
```

2. In - Order problem in Leet code

```python
from typing import Optional, List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def IOTWL(self,root,l):
        if root==None:
            return
        self.IOTWL(root.left,l)
        l.append(root.val)
        self.IOTWL(root.right,l)


    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        l=[]
        self.IOTWL(root,l)
        return l
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
sol = Solution()
in_order = sol.inorderTraversal(root)
print("In-order:", in_order)
```

3. Post - Order problem in Leet code

```python
from typing import Optional, List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def POTWL(self, root, l):
        if root is None:
            return
        self.POTWL(root.left, l)
        self.POTWL(root.right, l)
        l.append(root.val)

    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        l = []
        self.POTWL(root, l)
        return l
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
sol = Solution()
post_order = sol.postorderTraversal(root)
print("Post-order:", post_order)
```

4. Count complete Tree Nodes

```python
from typing import Optional, List

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def Count(self,root):
        if root == None:
            return 0
        l = self.Count(root.left)
        r = self.Count(root.right)
        return l + r + 1
    def countNodes(self, root: Optional[TreeNode]) -> int:
        list = []
        res = self.Count(root)
        return res
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
sol = Solution()
count = sol.countNodes(root)
print("Count:", count)
```
