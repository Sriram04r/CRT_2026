**BINARY SEARCH TRESS - (03-07-2026)**

1. BST is also a Binary Tree principles.

2. The values of left sub tree is lesser than root value.

3. The values of right sub tree is larger than root value.

4. It doesn't contains duplicate elements.

5. The sub trees should applicable BST properties.

WHENEVER A LEFT SUB TREE AND RIGHT SUB TREE EXISTS, DELETION OPERATION SHOULD PERFORM ON ROOT.THE NODE SHOULD SHOULD REPLACE ON LEFT SUB TREE PREDECESSOR AND RIGHT SUB TREE SUCCESSOR

**PROBLEMS - (LEET CODE)**

1. Search in a Binary search Tree

```python
class TreeNode:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None
class Solution:
    def searchBST(self, root, val):
        curr = root
        while curr != None:
            if curr.data == val:
                return curr
            elif val < curr.data:
                curr = curr.left
            else:
                curr = curr.right
        return None

root = TreeNode(4)
root.left = TreeNode(2)
root.right = TreeNode(7)
root.left.left = TreeNode(1)
root.left.right = TreeNode(3)

obj = Solution()

val = 99
result = obj.searchBST(root, val)

if result:
    print(f"The node exists in the Tree is: {result.data}")
else:
    print(f"The value {val} not exists in a Tree.")

```
