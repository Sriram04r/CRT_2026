```python
def max(a,b):
  if a > b:
    return a
  else:
    return b
def max(a,b,c):
  if a > b:
    return a
  elif b > c:
    return b
  else:
      return c

a = max(1,2,3)
print(a)
```

**ABSTRACTION**

```python
from abc import ABC,abstractmethod

class car1(ABC):
    @abstractmethod
    def brake(self):
        pass
    @abstractmethod
    def acceleration(self):
        pass
class car2(car1):
    def brake(self):
        print("brake enabled...")
    def acceleration(self):
        print("accelerating")

obj=car2()
obj.acceleration()
```
