1. Pre - Order Traversal - Root -> Left -> Right(Data printinting)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class preordertrav:
    def __init__(self):
        self.data = None

    def rootnode(self):
        self.data = Node(5)

    def printall(self,current_node):
        if current_node is not None:
            print(current_node.data,end = " ")
            self.printall(current_node.left)
            self.printall(current_node.right)

root = preordertrav()

root.rootnode()
root.data.left = Node(2)
root.data.left.left = Node(10)
root.data.left.left.left = Node(11)
root.data.right = Node(8)
root.data.right.right = Node(3)
root.data.right.right.left = Node(9)
root.data.right.right.right = Node(1)
root.printall(root.data)
```

```python
class TreeNode:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
def preOrder(root):
    if root==None:
        return
    print(root.data,end=" ")
    preOrder(root.left)
    preOrder(root.right)
root=TreeNode(1)
root.left=TreeNode(2)
root.right=TreeNode(3)
root.left.left=TreeNode(4)
root.left.right=TreeNode(5)
root.right.left=TreeNode(6)
root.right.right=TreeNode(7)

preOrder(root)
```

**2. In - Order Traversal -** Left -> Root -> Right

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class inordertrav:
    def __init__(self):
        self.data = None

    def rootnode(self):
        self.data = Node(5)

    def printall(self, current_node):
        if current_node is not None:
            self.printall(current_node.left)
            print(current_node.data, end=" ")
            self.printall(current_node.right)


root = inordertrav()

root.rootnode()
root.data.left = Node(2)
root.data.left.left = Node(10)
root.data.left.left.left = Node(11)
root.data.right = Node(8)
root.data.right.right = Node(3)
root.data.right.right.left = Node(9)
root.data.right.right.right = Node(1)

root.printall(root.data)
```

3. Post - Order Traversal - Left -> Right -> Root

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class postordertrav:
    def __init__(self):
        self.data = None

    def rootnode(self):
        self.data = Node(5)

    def printall(self, current_node):
        if current_node is not None:
            self.printall(current_node.left)
            self.printall(current_node.right)
            print(current_node.data, end=" ")


root = postordertrav()

root.rootnode()
root.data.left = Node(2)
root.data.left.left = Node(10)
root.data.left.left.left = Node(11)
root.data.right = Node(8)
root.data.right.right = Node(3)
root.data.right.right.left = Node(9)
root.data.right.right.right = Node(1)

root.printall(root.data)
```

**2. Breadth First Search -** Level order Traversal

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class Tree:
    def __init__(self):
        self.data = None

    def rootnode(self):
        self.data = Node(5)

    def preorder(self, node):
        if node:
            print(node.data, end=" ")
            self.preorder(node.left)
            self.preorder(node.right)

    def inorder(self, node):
        if node:
            self.inorder(node.left)
            print(node.data, end=" ")
            self.inorder(node.right)

    def postorder(self, node):
        if node:
            self.postorder(node.left)
            self.postorder(node.right)
            print(node.data, end=" ")


root = Tree()

root.rootnode()
root.data.left = Node(2)
root.data.left.left = Node(10)
root.data.left.left.left = Node(11)
root.data.right = Node(8)
root.data.right.right = Node(3)
root.data.right.right.left = Node(9)
root.data.right.right.right = Node(1)

print("Preorder:")
root.preorder(root.data)

print("\nInorder:")
root.inorder(root.data)

print("\nPostorder:")
root.postorder(root.data)
```
