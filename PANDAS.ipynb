```python
# Concatanation

import pandas as pd
import numpy as np
d1 = {
    'S_ID' : [111, 222, 333, 444, 555],
    'S_NAME' : ['Pspk', 'Rc', 'MB', 'AA', 'Pbs']

}

d2 = {
    'Branch' : ['CSM', 'CAI', 'CSD', 'AID', 'CSC'],
    'Location' : ['AP', 'TN', 'KL', 'KN', 'GJ']
}
ds1 = pd.DataFrame(d1)
ds2 = pd.DataFrame(d2)
ds3 = pd.concat([ds1, ds2], axis = 1)

print(ds3)
```

**JOINS**

There are mainly four types. They are
1. Inner join
2. Outer join
3. Left join
4. Right join

**Dumping** - The code is convert into file format.

**Pickle** - The file format is change into code.

```python
import pandas as pd
import numpy as np
d1= {
    'sid' : [111, 222, 333, 444, 555],
    'sName' : ['Pbs','Pspk', 'MB', 'AA','Rc' ],
    'branch' : ['CSD','CSM', 'AID', 'CSC','CAI']
}
d2= {
    'sid' : [444, 555, 666, 777, 888],
    'sname' : ['ANUSHKA', 'RSK', 'SP', 'KA', 'SMT'],
    'branch' : ['CSM', 'CAI', 'CSD', 'AID', 'CSC']
}

ds1=pd.DataFrame(d1)
ds2=pd.DataFrame(d2)
inner=pd.merge(ds1,ds2,on="sid",how="inner")
left=pd.merge(ds1,ds2,on="sid",how="left")
right=pd.merge(ds1,ds2,on="sid",how="right")
inner.to_csv("F:\\data science\\inner.csv")
left.to_csv("F:\\data science\\left.csv")
right.to_csv("F:\\data science\\right.csv")
print("files saved successfully!")
```
