```python
grid=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
    ]
n=len(grid)
m=len(grid[0])
sum=0
for i in range(n):
    for j in range(m):
        if i==0 or j==0 or i==n-1 or j==m-1:
            sum+=grid[i][j]
print(sum)
```

```python
class Solution:
    def maximumWealth(self, accounts):
        n = len(accounts)
        m = len(accounts[0])
        for i in range(n):
            maxx = 0
            for i in range(n):
                sum = 0
                for j in range(m):
                    sum += accounts[i][j]
                maxx = max(maxx, sum)
            return maxx
obj = Solution()
test_accounts = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
result = obj.maximumWealth(test_accounts)
print("Maximum Wealth:", result)
```
