1. Happy Number - Leet code

```python
class Solution:
    def sumOfDigits(self,n):
        sum = 0
        while n != 0:
            temp = n % 10
            sum = sum + temp ** 2
            n = n // 10
        return sum
    def isHappy(self, n: int) -> bool:
        s = set()
        while True:
            if n not in s:
                s.add(n)
                n = self.sumOfDigits(n)
            else:
                if n == 1:
                    return True
                else:
                    return False
sol = Solution()
sol.isHappy(12)
```

2. Linked List Cycle

```python

class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    def hasCycle(self, head):
        if head is None or head.next is None:
            return False
        f = head
        s = head
        while f and f.next:
            f = f.next.next
            s = s.next
            if f == s:
                return True
        return False

node1 = ListNode(1)
node2 = ListNode(2)
node3 = ListNode(3)
node4 = ListNode(4)

node1.next = node2
node2.next = node3
node3.next = node4


sol = Solution()
print(f"Has cycle: {sol.hasCycle(node1)}")

node4.next = node2
print(f"Has cycle after linking: {sol.hasCycle(node1)}")
```
