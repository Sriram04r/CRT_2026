```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import pandas as pd

data = {
    "house_name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "house_number": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "b": ["two", "one", "three", "one", "one", "two", "two", "two", "three", "three"],
    "h": [2, 1, 2, 3, 1, 2, 2, 1, 3, 1],
    "k": [2, 1, 1, 1, 1, 12, 2, 1, 1, 1],
    "age": [4, 3, 6, 7, 9, 4, 3, 2, 1, 2],
    "price": [55, 65, 70, 68, 67, 62, 60, 40, 45, 50]
}

ds = pd.DataFrame(data)
ds=ds.drop(columns=["house_name","house_number"])
ds["b"]=ds["b"].map({"one":1,"two":2,"three":3})
print(ds)
x=ds[["b","h","k","age"]]
y=ds[["price"]]
ds = pd.DataFrame(data)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

model=LinearRegression()
model.fit(x_train,y_train)

data2={
        "b": [2, 1, 3,1,1, 2,2,2,3,3],
        "h": [2, 1, 2, 3, 1, 2, 2, 1, 3, 1],
        "k": [2, 1, 1, 1, 1, 12, 2, 1, 1, 1],
        "age": [4, 3, 6, 7, 9, 4, 3, 2, 1, 2],
}
ds2=pd.DataFrame(data2)
model_op=model.predict(ds2)
print(model_op)
plt.scatter([2, 1, 2, 3, 1, 2, 2, 1, 3, 1],model_op,color="black")
# plt.plot([45,95,92,80,70,65],model_op,marker="o",color="r")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import pandas as pd
data={
    "student_hours":[20,30,2,10,35,15,12,13,18,4],
    "student_result":["pass","fail","pass","pass","pass","fail","pass","fail","pass","fail"],
}
ds = pd.DataFrame(data)
ds["student_result"]=ds["student_result"].map({"pass":1,"fail":0})
x=ds[["student_hours"]]
y=ds[["student_result"]]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LogisticRegression()
model.fit(x_train,y_train)
data2={
    "student_hours":[15,20,7,6,5]
}
sample_inp_ds=pd.DataFrame(data2)

model_op=model.predict(sample_inp_ds)
plt.scatter([15,20,7,6,5],model_op)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import pandas as pd

import pandas as pd
data={
    'Email':['mail1@gmail.com', 'mail2@gmail.com', 'mail3@gmail.com', 'mail4@gmail.com', 'mail5@gmail.com', 'mail6@gmail.com', 'mail7@gmail.com', 'mail8@gmail.com', 'mail9@gmail.com', 'mail10@gmail.com'],
    'Words':[100, 80, 40, 50, 60, 70, 80, 90, 20, 10],
    'characters' : [50, 20, 20, 50, 30, 40, 30, 60, 20, 50],
    'Links' : [4, 40, 10, 4, 2, 8, 12, 20, 36, 3],
    'Report' : [100, 20, 5, 150, 2, 50, 100, 30, 20, 150],
    'Label' : ['Spam', 'Spam', 'Not Spam', 'Spam', 'Not Spam', 'Spam', 'Spam', 'Not Spam', 'Not Spam', 'Spam']
}
ds = pd.DataFrame(data)

ds=ds.drop(columns=['Email'])
ds["Label"]=ds["Label"].map({"Spam":1,"Not Spam":0})

x=ds[["Words","characters","Links","Report"]]
y=ds[["Label"]]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

model=LogisticRegression()
model.fit(x_train,y_train)

data2={
    "Words":[70,80,90],
    "characters":[100,300,400],
    "Links":[6,8,19],
    "Report":[50,70,150]
}
sample_ds=pd.DataFrame(data2)
model_op=model.predict(sample_ds)
print(model_op)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import pandas as pd

import pandas as pd
data={
    'Email':['mail1@gmail.com', 'mail2@gmail.com', 'mail3@gmail.com', 'mail4@gmail.com', 'mail5@gmail.com', 'mail6@gmail.com', 'mail7@gmail.com', 'mail8@gmail.com', 'mail9@gmail.com', 'mail10@gmail.com'],
    'Words':[100, 80, 40, 50, 60, 70, 80, 90, 20, 10],
    'characters' : [50, 20, 20, 50, 30, 40, 30, 60, 20, 50],
    'Links' : [4, 40, 10, 4, 2, 8, 12, 20, 36, 3],
    'Report' : [100, 20, 5, 150, 2, 50, 100, 30, 20, 150],
    'Label' : ['Spam', 'Spam', 'Not Spam', 'Spam', 'Not Spam', 'Spam', 'Spam', 'Not Spam', 'Not Spam', 'Spam']
}
ds = pd.DataFrame(data)

ds=ds.drop(columns=['Email'])
ds["Label"]=ds["Label"].map({"Spam":1,"Not Spam":0})

x=ds[["Words","characters","Links","Report"]]
y=ds[["Label"]]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

model=LogisticRegression()
model.fit(x_train,y_train)

data2={
    "Words":[70,80,90],
    "characters":[100,300,400],
    "Links":[6,8,19],
    "Report":[50,70,150]
}
sample_ds=pd.DataFrame(data2)
model_op=model.predict(sample_ds)
print(model_op)
fig=plt.figure(figsize=(10,7))
ax=fig.add_subplot(111,projection="3d")

scatter=ax.scatter(ds["Words"],ds["characters"],ds["Links"],c=ds["Label"],s=60)
ax.set_title("3D email spam detection")
ax.set_xlabel("words")
ax.set_ylabel("characters")
ax.set_zlabel("links")
plt.show()
```
